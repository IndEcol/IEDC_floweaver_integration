```python
import pandas as pd
flows       = pd.read_excel('1_F_Wood_Products_Germany_2019_Szichta.xlsx','Data')
material    = 'wood, wood waste, and wood products'
layer       = 'dry mass'
region      = 'Germany'
scenario    = 'History'
time        = 2019

ExtractData = flows.loc[(flows['aspect 2 : material'] == material) & (flows['aspect 3 : layer'] == layer) & (flows['aspect 4 : region'] == region) & (flows['aspect 8 : scenario'] == scenario) & (flows['aspect 5 : time'] == time)]
SankeyDatar = ExtractData.drop(columns=['id','dataset_name','aspect 2 : material','aspect 3 : layer','aspect 4 : region','aspect 8 : scenario','aspect 5 : time','aspect 9 : ','aspect 10 : ','aspect 11 : ','aspect 12 : ','unitcode','unitcode1','stats_array_1','stats_array_2','stats_array_3','stats_array_4','comment','reserve1','reserve2','reserve3'])
SankeyDatar.rename(columns={'aspect 6 : origin_process': 'source', 'aspect 7 : destination_process': 'target','aspect 1 : commodity': 'type'}, inplace=True)
column_names= ['source','target','type','value']
SankeyDatar = SankeyDatar.reindex(columns=column_names)
SankeyDatar.reset_index(drop=True, inplace=True)
SankeyDatar
```

```python
from floweaver import *
import pandas as pd
Sankey_version = '1_F_Wood_Products_Germany_2019_Szichta'

# Define the processes
nodes = {
    'harvesting of wood': ProcessGroup(['harvesting of wood'], title='harvesting of wood'),
    'import of roundwood': ProcessGroup(['import of roundwood'], title='import of roundwood'),
    'export of roundwood': ProcessGroup(['export of roundwood'], title='export of roundwood'),
    'use of roundwood, unspecified': ProcessGroup(['use of roundwood, unspecified'], title='use of roundwood, unspecified'),
    'market for harvested wood': ProcessGroup(['market for harvested wood'], title='market for harvested wood'),
    'sawmill industry': ProcessGroup(['sawmill industry'], title='sawmill industry'),
    'production of final products and commodities': ProcessGroup(['production of final products and commodities'], title='final production'),
    'import of semi-finished products': ProcessGroup(['import of semi-finished products'], title='imports'),
    "export of semi-finished products": ProcessGroup(["export of semi-finished products"],title='exports'),
    'market for finished products': ProcessGroup(['market for finished products'], title='market for finished products'),
    "import of finished products": ProcessGroup(["import of finished products"],title='imports'),
    'export of finished products': ProcessGroup(['export of finished products'], title='exports'),
    'use phase - construction': ProcessGroup(['use phase - construction'], title='use phase - construction'),
    'use phase - furniture': ProcessGroup(['use phase - furniture'], title='use phase - furniture'),
    'use phase - paper and packaging': ProcessGroup(['use phase - paper and packaging'], title='use phase - paper and packaging'),
    'use phase - other': ProcessGroup(['use phase - other'], title='use phase - other'),
    'use phase': ProcessGroup(['use phase'], title='use phase'),
    'use of recovered wood, unspecified': ProcessGroup(['use of recovered wood, unspecified'], title='use of recovered wood, unspecified'),
    'treatment of industrial waste': ProcessGroup(['treatment of industrial waste'], title='treatment of industrial waste'),
    'market for sawmill byproducts': ProcessGroup(['market for sawmill byproducts'], title='market for sawmill byproducts'),
    'use of sawmill by-products and industrial residues, unspecified': ProcessGroup(['use of sawmill by-products and industrial residues, unspecified'], title='use of sawmill by-products and industrial residues, unspecified'),
    "loop1": Waypoint(title='', direction='L'),
    "loop2": Waypoint(title='', direction='L'),    
    "loop3": Waypoint(title='', direction='L'),
    "forward1": Waypoint(title='', direction='R'),    
}

# Define flows
bundles = [
    Bundle('import of roundwood', 'market for harvested wood'),
    Bundle('harvesting of wood', 'market for harvested wood'),
    Bundle('market for harvested wood', 'use of roundwood, unspecified'),
    Bundle('market for harvested wood', 'sawmill industry'),
    Bundle('market for harvested wood', 'export of roundwood'),
    Bundle('sawmill industry','production of final products and commodities'),
    Bundle('sawmill industry', 'market for sawmill byproducts', waypoints=['forward1']),
    Bundle('import of semi-finished products','production of final products and commodities'),
    Bundle('production of final products and commodities','export of semi-finished products'),
    Bundle('production of final products and commodities', 'market for finished products'),  
    Bundle('production of final products and commodities', 'market for sawmill byproducts'),
    Bundle('import of finished products', 'market for finished products'), 
    Bundle('market for finished products', 'export of finished products'),    
    Bundle('market for finished products', 'use phase - construction'),
    Bundle('market for finished products', 'use phase - furniture'),
    Bundle('market for finished products', 'use phase - paper and packaging'),
    Bundle('market for finished products', 'use phase - other'),
    Bundle('use phase - construction', 'use phase'),
    Bundle('use phase - furniture', 'use phase'),
    Bundle('use phase - paper and packaging', 'use phase'),
    Bundle('use phase - other', 'use phase'),
    Bundle('use phase', 'treatment of industrial waste'),
    Bundle('use phase', 'use of recovered wood, unspecified'),
    Bundle('use phase', 'sawmill industry', waypoints=['loop2']),
    Bundle('sawmill industry', 'sawmill industry', waypoints=['loop1'], flow_selection='type == "wood products, semi-finished"'), 
    Bundle('market for sawmill byproducts', 'use of sawmill by-products and industrial residues, unspecified'),    
    Bundle('market for sawmill byproducts', 'sawmill industry', waypoints=['loop3']),
]

# Set the colours for the labels in the partition.
palette = {
    'coniferous wood': 'lightgreen',
    'non-coniferous wood': 'olivedrab',
    'bark': 'darkgoldenrod',
    'roundwood': 'yellowgreen',
    'sawmill by-products': 'lightgrey',
    'wood products, semi-finished': 'wheat',
    'wood products': 'lightblue',
    'wood waste': 'tan',
    'production residue': 'silver',
    'wood, recovered': 'tan',
}

#Define the ordering of the processes and waypoints
ordering = [
    [["import of roundwood"],["harvesting of wood"],[],[],[]],
    [[],["market for harvested wood"],[],[],[]],
    [["import of semi-finished products"],["sawmill industry",'loop1'],['export of roundwood'],[],['use of roundwood, unspecified']],
    [["import of finished products"],["production of final products and commodities"],['forward1'],[],[]],
    [[],['market for finished products'],['market for sawmill byproducts'],['loop3'],['export of semi-finished products']],
    [[],['use phase - construction','use phase - furniture','use phase - paper and packaging','use phase - other'],[],[],['export of finished products']],
    [[],['use phase'],[],['loop2'],['use of sawmill by-products and industrial residues, unspecified']],
    [[],['use of recovered wood, unspecified'],['treatment of industrial waste'],[],[]],
]

# Generate the Sankey diagram
size = dict(width=1400, height=400)

# New SDD with the flow_partition set
sdd = SankeyDefinition(nodes, bundles, ordering, flow_partition=Partition.Simple('type', list(palette.keys())))

Sankey = weave(sdd, SankeyDatar, palette=palette).to_widget(**size)

Sankey # Show below

Sankey.auto_save_png(Sankey_version + '_' + region + '_' + str(time) + '.png')
Sankey.auto_save_svg(Sankey_version + '_' + region + '_' + str(time) + '.svg')
```

```python
# double-check and export Sankey specification as .json
if isinstance(SankeyDatar, pd.DataFrame):
    SankeyDatar = Dataset(SankeyDatar)
    
spec = compile_sankey_definition(
    sdd,       # see above: sdd = SankeyDefinition(nodes, bundles, ordering, flow_partition=flows_by_type)
    measures   = "value",
    link_width = None,
    link_color = None,
    palette    = palette,
    add_elsewhere_waypoints = True,
    dimension_tables        = None,
)

Sankey = execute_weave(spec, SankeyDatar)
Sankey.to_widget(**size)
```

```python
# export Sankey spec as .json for use in IEDC web application
import json
json_data = json.dumps(spec.to_json(), indent=2)
# Write Sankey spec to .json, filename as indicated in: 
# https://github.com/IndEcol/IEDC_floweaver_integration/blob/main/floweaver_IEDC_Sankey_configs/floweaver_IEDC_Sankey_configs.config
with open(Sankey_version + '.json', "w", encoding="utf-8") as f:
    f.write(json_data)
    f.write("\n")
```
