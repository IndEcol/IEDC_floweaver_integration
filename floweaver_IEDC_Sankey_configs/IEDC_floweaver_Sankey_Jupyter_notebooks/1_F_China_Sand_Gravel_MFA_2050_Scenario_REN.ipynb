```python
import pandas as pd
flows       = pd.read_excel('1_F_China_Sand_Gravel_MFA_2050_Scenario_REN.xlsx','Data')
material    = 'construction minerals, related materials and waste'
layer       = 'dry mass'
region      = 'China'
scenario    = 'all measures'
time        = 2050

ExtractData = flows.loc[(flows['aspect 2 : material'] == material) & (flows['aspect 3 : layer'] == layer) & (flows['aspect 4 : region'] == region) & (flows['aspect 8 : scenario'] == scenario) & (flows['aspect 5 : time'] == time)]
SankeyDatar = ExtractData.drop(columns=['id','dataset_name','aspect 2 : material','aspect 3 : layer','aspect 4 : region','aspect 8 : scenario','aspect 5 : time','aspect 9 : ','aspect 10 : ','aspect 11 : ','aspect 12 : ','unitcode','unitcode1','stats_array_1','stats_array_2','stats_array_3','stats_array_4','comment','reserve1','reserve2','reserve3'])
SankeyDatar.rename(columns={'aspect 6 : origin_process': 'source', 'aspect 7 : destination_process': 'target','aspect 1 : commodity': 'type'}, inplace=True)
column_names= ['source','target','type','value']
SankeyDatar = SankeyDatar.reindex(columns=column_names)
SankeyDatar.reset_index(drop=True, inplace=True)
SankeyDatar
```

```python
import pandas as pd
from floweaver import *

Sankey_version = 'China_Sand_Gravel_REN_Sankey_as_in_Paper'

# Set the default size to fit the documentation better.
size = dict(width=900, height=400)
SCALE = 7.0

nodes = {
    'sources':    ProcessGroup(['mining - natural sand and gravel',
                                'mining - crushed stone',
                                'market for recycled materials']),
    'construction sector': ProcessGroup(['building and construction']),
    'final_use' : ProcessGroup(['use phase - buildings',
                                'use phase - infrastructure']),
    'waste management' : ProcessGroup(['waste management']),
    'finalsink' : ProcessGroup(['landfilling',
                                'building and construction - use of fillings',
                                'recycling - aggregates']),
    'recycled aggregates'      : Waypoint(direction='L')}


ordering = [
    ['sources'],       # put "sources" on the left...
    ['construction sector'],
    ['final_use','recycled aggregates'],
    ['waste management'],
    ['finalsink'],   # ... and "finalsink" on the right.
]


bundles = [
    Bundle('sources',    'construction sector'),
    Bundle('construction sector', 'final_use'),
    Bundle('final_use',  'waste management'),
    Bundle('construction sector', 'waste management'),    
    Bundle('waste management',  'finalsink'),
    Bundle('finalsink', 'sources', waypoints=['recycled aggregates'])]


nodes['sources'].partition   = Partition.Simple('process', ['mining - natural sand and gravel',
                                                            'mining - crushed stone',
                                                            'market for recycled materials'])
nodes['final_use'].partition = Partition.Simple('process', ['use phase - buildings', 'use phase - infrastructure'])
nodes['finalsink'].partition = Partition.Simple('process', ['landfilling',
                                                            'building and construction - use of fillings',
                                                            'recycling - aggregates'])


# Partition by the "type/commodity" column of the flows table
flows_by_type = Partition.Simple('type', ['waste, construction, mineralic, non-metallic', 'recycled aggregates', 
                                          'waste, demolition - buildings, mineralic, non-metallic',
                                          'waste, demolition - infrastructure, mineralic, non-metallic',
                                          'recycled aggregates from construction waste',
                                          'recycled aggregates from demolished buildings',
                                          'recycled aggregates from demolished infrastructure',
                                          'aggregates', 'sand and gravel', 'crushed stone'])

# Set the colours for the labels in the partition.
palette = {'waste, construction, mineralic, non-metallic': 'greenyellow','recycled aggregates': 'yellow',
           'recycled aggregates from construction waste': 'greenyellow',
           'aggregates': 'purple', 'sand and gravel': 'chocolate', 'crushed stone': 'orange',
           'waste, demolition - buildings, mineralic, non-metallic': 'darkgreen',
           'waste, demolition - infrastructure, mineralic, non-metallic':'limegreen',
           'recycled aggregates from demolished buildings': 'darkgreen',
           'recycled aggregates from demolished infrastructure':'limegreen'}

# New SDD with the flow_partition set
sdd = SankeyDefinition(nodes, bundles, ordering, 
                       flow_partition=flows_by_type)

# w1 = weave(sdd, flows, palette=palette).to_widget(**size).auto_save_png('sand_example_v1.png')
Sankey = weave(sdd, SankeyDatar, palette=palette).to_widget(**size)
Sankey.scale = SCALE

#Sankey # Show below
Sankey.auto_save_png(Sankey_version + '_' + region + '_' + str(time) + '.png')
Sankey.auto_save_svg(Sankey_version + '_' + region + '_' + str(time) + '.svg')
```

```python
# double-check and export Sankey specification as .json
if isinstance(SankeyDatar, pd.DataFrame):
    SankeyDatar = Dataset(SankeyDatar)
    
spec = compile_sankey_definition(
    sdd,       # see above: sdd = SankeyDefinition(nodes, bundles, ordering, flow_partition=flows_by_type)
    measures   = "value",
    link_width = None,
    link_color = None,
    palette    = palette,
    add_elsewhere_waypoints = True,
    dimension_tables        = None,
)

Sankey = execute_weave(spec, SankeyDatar)
Sankey.to_widget(**size)
```

```python
# export Sankey spec as .json for use in IEDC web application
import json
json_data = json.dumps(spec.to_json(), indent=2)
# Write Sankey spec to .json, filename as indicated in: 
# https://github.com/IndEcol/IEDC_floweaver_integration/blob/main/floweaver_IEDC_Sankey_configs/floweaver_IEDC_Sankey_configs.config
with open(Sankey_version + '.json', "w", encoding="utf-8") as f:
    f.write(json_data)
    f.write("\n")
```
