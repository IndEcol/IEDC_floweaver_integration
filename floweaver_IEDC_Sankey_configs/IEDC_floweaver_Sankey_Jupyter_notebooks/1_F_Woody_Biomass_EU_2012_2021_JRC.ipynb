```python
import pandas as pd
flows       = pd.read_excel('1_F_woody_biomass_EU_2012_2021_JRC.xlsx','Data')
material    = 'wood and wood products, related materials and waste'
layer       = 'dry mass'
region      = 'Germany'
#region      = 'Finland'
scenario    = 'History'
time        =  2020

ExtractData = flows.loc[(flows['aspect 2 : material'] == material) & (flows['aspect 3 : layer'] == layer) & (flows['aspect 4 : region'] == region) & (flows['aspect 8 : scenario'] == scenario) & (flows['aspect 5 : time'] == time)]
SankeyDatar = ExtractData.drop(columns=['id','dataset_name','aspect 2 : material','aspect 3 : layer','aspect 4 : region','aspect 8 : scenario','aspect 5 : time','aspect 9 : ','aspect 10 : ','aspect 11 : ','aspect 12 : ','unitcode','unitcode1','stats_array_1','stats_array_2','stats_array_3','stats_array_4','comment','reserve1','reserve2','reserve3'])
SankeyDatar.rename(columns={'aspect 6 : origin_process': 'source', 'aspect 7 : destination_process': 'target','aspect 1 : commodity': 'type'}, inplace=True)
column_names= ['source','target','type','value']
SankeyDatar = SankeyDatar.reindex(columns=column_names)
SankeyDatar.reset_index(drop=True, inplace=True)
SankeyDatar
```

```python
from floweaver import *

# Define the Sankey version, which will be the name of the .json spec file. 
# Default: Sankey version = IEDC template filename without 1_F_ and without .xlxs, start with capital letter
Sankey_version = 'Woody_biomass_EU_2012_2021_JRC'

nodes = {
    'forestry': ProcessGroup(['forestry'], title='forestry'),
    'market for harvested wood': ProcessGroup(['market for harvested wood'], title='market for harvested wood'),
    'import of wood': ProcessGroup(['import of wood'], title='imports'),
    'export of roundwood': ProcessGroup(['export of roundwood'], title='exports'),
    'processing of residual bark': ProcessGroup(['processing of residual bark'], title='processing of residual bark'),
    'wood waste supply': ProcessGroup(['wood waste supply'], title='wood waste supply'),
    'market for fuelwood and wood residuals': ProcessGroup(['market for fuelwood and wood residuals'], title='market for fuelwood and wood residuals'),
    'sawmill industry': ProcessGroup(['sawmill industry'], title='sawmill industry'),
    'panel industry': ProcessGroup(['panel industry'], title='panel industry'),
    'pulp industry': ProcessGroup(['pulp industry'], title='pulp industry'),
    'market for fibre boards': ProcessGroup(['market for fibre boards'], title='market for fibre boards'),
    'market for plywood and veneer sheets': ProcessGroup(['market for plywood and veneer sheets'], title='market for plywood and veneer sheets'),
    'market for particle boards': ProcessGroup(['market for particle boards'], title='market for particle boards'),
    'import of fibre board': ProcessGroup(['import of fibre board'], title='imports'),
    'export of fibre board': ProcessGroup(['export of fibre board'], title='exports'),
    'import of particle board': ProcessGroup(['import of particle board'], title='imports'),
    'export of particle board': ProcessGroup(['export of particle board'], title='exports'),
    'import of plywood and veneer sheets': ProcessGroup(['import of plywood and veneer sheets'], title='imports'),
    'export of plywood and veneer sheets': ProcessGroup(['export of plywood and veneer sheets'], title='export'),
    'import of sawnwood': ProcessGroup(['import of sawnwood'], title='imports'),
    'market for sawnwood': ProcessGroup(['market for sawnwood'], title='market for sawnwood'),
    'export of sawnwood': ProcessGroup(['export of sawnwood'], title='exports'),
    'import of pulp': ProcessGroup(['import of pulp'], title='imports'),
    'import of recovered pulp': ProcessGroup(['import of recovered pulp'], title='imports'),
    'market for paper pulp': ProcessGroup(['market for paper pulp'], title='market for paper pulp'),
    'export of pulp': ProcessGroup(['export of pulp'], title='exports'),
    'market for dissolving pulp': ProcessGroup(['market for dissolving pulp'], title='market for dissolving pulp'),
    'import of dissolving pulp': ProcessGroup(['import of dissolving pulp'], title='imports'),
    'export of dissolving pulp': ProcessGroup(['export of dissolving pulp'], title='exports'),
    'market for fuel wood and wood processing by-products': ProcessGroup(['market for fuel wood and wood processing by-products'], title='by-products'),
    'other supply of wood processing by-products': ProcessGroup(['other supply of wood processing by-products'], title='other supply'),
    'import of wood processing by-products': ProcessGroup(['import of wood processing by-products'], title='imports'),
    'export of wood processing by-products': ProcessGroup(['export of wood processing by-products'], title='exports'),
    'wood pellets industry': ProcessGroup(['wood pellets industry'], title='wood pellets industry'),
    'energy recovery': ProcessGroup(['energy recovery'], title='energy recovery'),
    'import of wood pellets': ProcessGroup(['import of wood pellets'], title='imports'),
    'export of wood pellets': ProcessGroup(['export of wood pellets'], title='exports'),
    "loop1": Waypoint(title='', direction='L'),
    "loop2": Waypoint(title='', direction='L'),
    "loop3": Waypoint(title='', direction='L'),
    "loop4": Waypoint(title='', direction='L'),
    "D5": Waypoint(title='', direction='R'),
    "D6": Waypoint(title='', direction='R'),
    "D7": Waypoint(title='', direction='R'),
    "D8": Waypoint(title='', direction='R'),
    "D9": Waypoint(title='', direction='R'),
    "D10": Waypoint(title='', direction='R'),
}
    
bundles = [
    Bundle('forestry', 'market for harvested wood'),
    Bundle('import of wood', 'market for harvested wood'),
    Bundle('wood waste supply', 'panel industry'),
    Bundle('wood waste supply', 'market for fuelwood and wood residuals'),
    Bundle('market for harvested wood', 'market for fuelwood and wood residuals'),
    Bundle('market for harvested wood', 'processing of residual bark'),
    Bundle('market for harvested wood', 'export of roundwood'),
    Bundle('market for harvested wood', 'panel industry', waypoints=['D5','D6']),
    Bundle('market for harvested wood', 'sawmill industry', waypoints=['D7','D8']),
    Bundle('market for harvested wood', 'pulp industry', waypoints=['D9','D10']),
    Bundle('panel industry', 'market for fibre boards'),
    Bundle('panel industry', 'market for plywood and veneer sheets'),
    Bundle('panel industry', 'market for particle boards'),
    Bundle('panel industry', 'market for fuel wood and wood processing by-products'),
    Bundle('import of particle board', 'market for particle boards'),  
    Bundle('market for particle boards','export of particle board'),
    Bundle('import of fibre board', 'market for fibre boards'), 
    Bundle('market for fibre boards', 'export of fibre board'), 
    Bundle('import of plywood and veneer sheets', 'market for plywood and veneer sheets'),
    Bundle('market for plywood and veneer sheets', 'export of plywood and veneer sheets'),
    Bundle('sawmill industry', 'market for fuel wood and wood processing by-products'),
    Bundle('sawmill industry', 'market for sawnwood'),
    Bundle('import of sawnwood','market for sawnwood'),
    Bundle('market for sawnwood','export of sawnwood'),
    Bundle('import of wood processing by-products', 'market for fuel wood and wood processing by-products'),
    Bundle('import of pulp','market for paper pulp'),
    Bundle('import of recovered pulp','market for paper pulp'),
    Bundle('market for paper pulp','export of pulp'),
    Bundle('pulp industry', 'market for fuel wood and wood processing by-products'),
    Bundle('pulp industry', 'market for paper pulp'),
    Bundle('pulp industry', 'market for dissolving pulp'),
    Bundle('import of dissolving pulp', 'market for dissolving pulp'),
    Bundle('market for dissolving pulp','export of dissolving pulp'),
    Bundle('market for fuel wood and wood processing by-products','market for fuelwood and wood residuals'),
    Bundle('market for fuel wood and wood processing by-products','export of wood processing by-products'), 
    Bundle('market for fuelwood and wood residuals', 'energy recovery'),
    Bundle('market for fuelwood and wood residuals', 'wood pellets industry'),
    Bundle('wood pellets industry', 'energy recovery'),
    Bundle('wood pellets industry', 'export of wood pellets'),
    Bundle('import of wood pellets', 'energy recovery'),
    Bundle('other supply of wood processing by-products', 'market for fuel wood and wood processing by-products'),
    Bundle('market for fuel wood and wood processing by-products','pulp industry', waypoints=['loop3','loop4'], flow_selection='type == "wood processing by-products"'),
    Bundle('market for fuel wood and wood processing by-products','panel industry', waypoints=['loop1','loop2'], flow_selection='type == "wood processing by-products"'),
]

ordering = [
    [['import of wood'],['forestry'],[]],
    [[],['market for harvested wood']],
    [[],['D5'],['D7'],['D9'],['export of roundwood'],['processing of residual bark']],
    [[],['wood waste supply'],['D6'],['D8'],[],[],['D10']],
    [[],['loop2'],['panel industry'],['import of plywood and veneer sheets'],['import of particle board'],['import of fibre board'],['sawmill industry'],['import of sawnwood'],['pulp industry'],[],['import of pulp'],[],['import of recovered pulp'],[],['import of dissolving pulp'],[],[],['loop4']],
    [[],[],['market for plywood and veneer sheets'],['market for particle boards'],['market for fibre boards'],[],['market for sawnwood'],[],[],[],[],['market for paper pulp'],[],[],[],[],['market for dissolving pulp'],[]],
    [[],[],['export of plywood and veneer sheets'],['export of particle board'],['export of fibre board'],['import of wood processing by-products'],['export of sawnwood'],[],[],['export of pulp'],[],[],[],[],['export of dissolving pulp'],[],[],[],[],['other supply of wood processing by-products']],
    [[],['loop1'],[],[],[],['market for fuel wood and wood processing by-products'],[],[],[],[],[],[],[],[],[],[],[],['loop3']],
    [[],['market for fuelwood and wood residuals'],[],[],[],[],['export of wood processing by-products']],
    [['import of wood pellets'],[],['wood pellets industry']],
    [[],['energy recovery'],['export of wood pellets']],
]

flows_by_type = Partition.Simple('type', [
    'wood',
    'bark',
    'roundwood',
    'wood processing by-products',
    'fuel wood and wood processing by-products',
    'particle board',
    'fibre board',
    'plywood and veneer',
    'sawnwood',
    'pulp',
    'dissolving pulp',
    'recovered pulp',
    'wood pellets',
    'post consumer wood',
])

palette = {
    'wood': 'forestgreen',
    'bark': 'darkgreen',
    'roundwood': 'limegreen',
    'wood processing by-products': 'orange',
    'fuel wood and wood processing by-products': 'dodgerblue',
    'particle board': 'tomato',
    'fibre board': 'tomato',
    'plywood and veneer': 'tomato',
    'sawnwood': 'tomato',
    'pulp': 'tomato',
    'dissolving pulp': 'tomato',
    'recovered pulp': 'tomato',
    'wood pellets': 'turquoise',
    'post consumer wood': 'gold',
}

# Specify canvas size and scale (thickness) of the flows:
size = dict(width=1400, height=450)
SCALE = 0.001

sdd = SankeyDefinition(nodes, bundles, ordering, flow_partition=flows_by_type)

Sankey = weave(sdd, SankeyDatar, palette=palette).to_widget(**size)
Sankey.scale = SCALE

Sankey

Sankey.auto_save_png(Sankey_version + '_' + region + '_' + str(time) + '.png')
Sankey.auto_save_svg(Sankey_version + '_' + region + '_' + str(time) + '.svg')
```

```python
# double-check and export Sankey specification as .json
if isinstance(SankeyDatar, pd.DataFrame):
    SankeyDatar = Dataset(SankeyDatar)
    
spec = compile_sankey_definition(
    sdd,       # see above: sdd = SankeyDefinition(nodes, bundles, ordering, flow_partition=flows_by_type)
    measures   = "value",
    link_width = None,
    link_color = None,
    palette    = palette,
    add_elsewhere_waypoints = True,
    dimension_tables        = None,
)

Sankey = execute_weave(spec, SankeyDatar)
Sankey.to_widget(**size)
```

```python
# export Sankey spec as .json for use in IEDC web application
import json
json_data = json.dumps(spec.to_json(), indent=2)
# Write Sankey spec to .json, filename as indicated in: 
# https://github.com/IndEcol/IEDC_floweaver_integration/blob/main/floweaver_IEDC_Sankey_configs/floweaver_IEDC_Sankey_configs.config
with open(Sankey_version + '.json', "w", encoding="utf-8") as f:
    f.write(json_data)
    f.write("\n")
```
